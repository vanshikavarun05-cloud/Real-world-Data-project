# Sample Superstore – Real-World Data Analysis
This notebook inspects the retail dataset and visualizes sales, profit and correlations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_FILE = Path('SampleSuperstore.csv')
OUTPUT_DIR = Path('outputs')
OUTPUT_DIR.mkdir(exist_ok=True)
sns.set_theme(style='whitegrid')

if not DATA_FILE.exists():
    raise FileNotFoundError('Place SampleSuperstore.csv in the same folder as this notebook.')

df = pd.read_csv(DATA_FILE, encoding='latin1')
print('Shape:', df.shape)
display(df.head())
df.info()
print('\nMissing values:\n', df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
display(df.describe())
```

## 1. Total sales by category

```python
sales_by_category = df.groupby('Category', as_index=False)['Sales'].sum().sort_values('Sales', ascending=False)
plt.figure(figsize=(8, 5))
sns.barplot(data=sales_by_category, x='Category', y='Sales', color='steelblue')
plt.title('Total Sales by Category')
plt.xlabel('Category'); plt.ylabel('Total Sales')
plt.tight_layout()
plt.savefig(OUTPUT_DIR / 'sales_by_category.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```

## 2. Total profit by region

```python
profit_by_region = df.groupby('Region', as_index=False)['Profit'].sum().sort_values('Profit', ascending=False)
plt.figure(figsize=(8, 5))
sns.barplot(data=profit_by_region, x='Region', y='Profit', color='steelblue')
plt.title('Total Profit by Region')
plt.xlabel('Region'); plt.ylabel('Total Profit')
plt.tight_layout()
plt.savefig(OUTPUT_DIR / 'profit_by_region.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```

## 3. Sales versus profit

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='Sales', y='Profit', hue='Category', alpha=0.55, s=35)
plt.title('Sales vs. Profit by Category')
plt.xlabel('Sales'); plt.ylabel('Profit')
plt.tight_layout()
plt.savefig(OUTPUT_DIR / 'sales_vs_profit.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```

## 4. Correlation heatmap
Identifier columns are excluded because their numeric codes do not represent measured business quantities.

```python
numeric_for_corr = df[['Sales', 'Quantity', 'Discount', 'Profit']]
correlation = numeric_for_corr.corr()
plt.figure(figsize=(7, 5))
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', center=0, linewidths=0.5)
plt.title('Correlation Heatmap of Sales Variables')
plt.tight_layout()
plt.savefig(OUTPUT_DIR / 'correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
print('Analysis complete. Charts saved in:', OUTPUT_DIR.resolve())
```
